```python

from pyspark.sql import SparkSession

# Initialize a simple Spark Session
spark = SparkSession.builder \
    .appName("My Docker Spark") \
    .getOrCreate()

df = (spark.read
      .option("header", True)
      .option("inferSchema", True)
      .csv("./fec_transactions.csv")
      )

df.select("transaction_dt", "amount_usd").show(5, truncate=False)

```
